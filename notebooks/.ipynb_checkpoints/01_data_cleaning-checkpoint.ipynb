## Data Grain Understanding

Understanding the level of detail (grain) of each table is critical to avoid incorrect joins and duplication.

- orders → one row per order  
- order_items → one row per item in an order  
- payments → one row per payment transaction (multiple per order possible)  
- customers → one row per customer  
- products → one row per product  
- reviews → one row per review  

### Key Insight
Joining tables with different grain (like order_items and payments) can create duplication.

Example:
1 order → 3 items → 29 payments  
Direct join → 3 × 29 = 87 rows (this will give incorrect data)

### Conclusion
Revenue should be calculated using **order_items** to avoid duplication errors.

# CLEANING TABLE ORDERS

```python
# Convert date columns
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_approved_at'] = pd.to_datetime(orders['order_approved_at'])
orders['order_delivered_carrier_date'] = pd.to_datetime(orders['order_delivered_carrier_date'])
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'], errors='coerce')
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])
```

```python
orders['is_delivered'] = orders['order_delivered_customer_date'].notnull()
```

# CLEANING ORDER_ITEMS

```python
order_items['price']=order_items['price'].astype(float)
```

```python
order_items['freight_value']=order_items['freight_value'].astype(float)
```

# HANDLING MISSING VALUE OF PRODUCTS TABLE

```python
products['product_category_name'] = products['product_category_name'].fillna('Unknown')
```

```python
orders.to_csv("C:/Users/MUSKAN/ecommerce-revenue-intelligence-customer-insights/data/processed/clean_orders.csv", index=False)
order_items.to_csv("C:/Users/MUSKAN/ecommerce-revenue-intelligence-customer-insights/data/processed/clean_order_items.csv", index=False)
payments.to_csv("C:/Users/MUSKAN/ecommerce-revenue-intelligence-customer-insights/data/processed/clean_payments.csv", index=False)
customers.to_csv("C:/Users/MUSKAN/ecommerce-revenue-intelligence-customer-insights/data/processed/clean_customers.csv", index=False)
products.to_csv("C:/Users/MUSKAN/ecommerce-revenue-intelligence-customer-insights/data/processed/clean_products.csv", index=False)
reviews.to_csv("C:/Users/MUSKAN/ecommerce-revenue-intelligence-customer-insights/data/processed/clean_reviews.csv", index=False)
```
